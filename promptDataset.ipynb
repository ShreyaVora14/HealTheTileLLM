```python
import pandas as pd
import numpy as np

# Full list of ailments and date ranges
health_ailments = [
    "Hypertension (High Blood Pressure)", "Type 2 Diabetes", "Asthma", "Anxiety Disorder",
    "Depression", "Osteoarthritis", "Chronic Back Pain", "Migraine", "Obesity",
    "High Cholesterol (Hyperlipidemia)", "Acid Reflux (GERD)", "Irritable Bowel Syndrome (IBS)",
    "Hypothyroidism", "Insomnia", "Allergic Rhinitis (Seasonal Allergies)",
    "Psoriasis Urinary Tract Infection (UTI)", "Anemia Eczema (Atopic Dermatitis)", "Common Cold",
    "Heart Palpitations", "Pneumonia", "Bradycardia (Slow Heart Rate)", "Tachycardia (Fast Heart Rate)",
    "Postural Orthostatic Tachycardia Syndrome (POTS)", "Heat Exhaustion", "Dehydration",
    "Overtraining Syndrome", "Shin Splints", "Runner’s Knee (Patellofemoral Pain Syndrome)",
    "Achilles Tendinitis", "Iliotibial Band Syndrome (ITBS)", "Exercise-Induced Hypoglycemia",
    "Muscle Strain", "Joint Sprain", "Exertional Rhabdomyolysis", "Fatigue from Iron Deficiency",
    "Hypertrophic Cardiomyopathy", "Delayed Onset Muscle Soreness (DOMS)", "Hypotension (Low Blood Pressure)",
    "Orthostatic Intolerance", "Myofascial Pain Syndrome", "Attention Deficit Hyperactivity Disorder (ADHD)",
    "Chronic Kidney Disease", "Rheumatoid Arthritis", "Gout", "Chronic Sinusitis", "Celiac Disease",
    "Fibromyalgia", "Chronic Fatigue Syndrome", "Glaucoma", "Epilepsy", "Peripheral Neuropathy",
    "Polycystic Ovary Syndrome (PCOS)", "Plantar Fasciitis", "Shingles (Herpes Zoster)",
    "Benign Prostatic Hyperplasia (BPH)", "Gallstones", "Restless Legs Syndrome", "Hyperthyroidism",
    "Rosacea", "Peptic Ulcer Disease"
]

date_ranges = [
    "Yesterday", "Last 36 hours", "Last 3 hours", "Past week", "Past 2 weeks", "Past 30 days",
    "Last month", "Last 2 months", "Last 3 months", "This month (up to today)",
    "03/01/2020 – 03/15/2020", "2019/11/29 – 2019/12/15", "12/30/2019 – 01/10/2020",
    "2019/11/05 – 2019/11/06", "03/20/2020 – 03/27/2020", "2020/02/05 – 2020/02/10",
    "Past 1 hour", "Past 3 hours", "Past 6 hours", "Past 12 hours", "Today, 08:00 AM – 11:00 AM",
    "2019/11/10 – 2019/11/12", "2019/11/18 – 2019/11/22", "2019/12/01 – 2019/12/05",
    "2019/12/08 – 2019/12/12", "2019/12/20 – 2019/12/27", "2020/01/01 – 2020/01/07",
    "2020/01/11 – 2020/01/15", "2020/01/20 – 2020/01/25", "2020/01/28 – 2020/02/02",
    "2020/02/11 – 2020/02/14", "2020/02/15 – 2020/02/20", "2020/02/22 – 2020/02/25",
    "03/01/2020 – 03/05/2020", "03/06/2020 – 03/10/2020", "03/11/2020 – 03/14/2020",
    "03/15/2020 – 03/19/2020", "03/21/2020 – 03/25/2020", "03/26/2020 – 03/29/2020",
    "2019/11/03 – 2019/11/07", "2019/12/15 – 2019/12/19", "Last weekend", "Since the beginning of this year",
    "The first week of November 2019", "Between 02/10/2020 and 02/20/2020", "For the past 4 days",
    "Since last Monday", "The entire month of February 2020", "During Spring 2020",
    "From 07:00 PM to 10:00 PM yesterday", "Over the last school semester", "Since my last doctor’s visit",
    "The last quarter of 2019", "Since turning 30 years old", "Every morning this week",
    "Since the COVID-19 pandemic began", "From 03/01/2020 to 03/07/2020", "During the holiday season 2019",
    "The second half of November 2019", "For the past 10 minutes", "Since last night at midnight"
]

formats = [
    "I have had {a} since {d}.",
    "I've been experiencing {a} since {d}.",
    "{a} have been affecting me since {d}.",
    "Since {d}, I’ve been dealing with {a}.",
    "I’ve been suffering from {a} ever since {d}.",
    "I have {a} since {d}."
]

# Generate 3000 randomized prompts
np.random.seed(0)
random_ailments = np.random.choice(health_ailments, 3000, replace=True)
random_dates = np.random.choice(date_ranges, 3000, replace=True)
random_formats = np.random.choice(formats, 3000, replace=True)

# Generate final prompts
prompts = [f.format(a=a, d=d) for f, a, d in zip(random_formats, random_ailments, random_dates)]

# Create DataFrame
df_all = pd.DataFrame({
    "prompt": prompts,
    "health_ailment": random_ailments,
    "date": random_dates
})

#  Save to CSV if needed
df_all.to_csv("health_prompts.csv", index=False)

#  Preview
print(df_all.head())
print(f"\nTotal prompts: {len(df_all)}")

```
